```python
batting_first=ipl[ipl["win_by_runs"]!=0]
```

```python
batting_first.head()
```

```python
plt.figure(figsize=(5,5))
plt.hist(batting_first["win_by_runs"])
plt.title("Distribution of runs")
plt.xlabel("Runs")
plt.show()
```

```python
batting_first["winner"].value_counts()
```

```python
plt.figure(figsize=(6,5))
plt.bar(list(batting_first["winner"].value_counts()[0:3].keys()),list(ipl["player_of_match"].value_counts()[0:3]),color=["blue","yellow","orange"])
plt.show
```

```python
plt.figure(figsize=(6,5))
plt.pie(batting_first["winner"].value_counts(),labels=list(batting_first["winner"].value_counts().keys()),autopct="%.1f%%")
plt.show
```
